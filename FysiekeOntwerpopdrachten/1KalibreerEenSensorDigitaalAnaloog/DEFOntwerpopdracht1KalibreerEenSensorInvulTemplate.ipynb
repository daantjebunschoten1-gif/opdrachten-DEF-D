## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit


Gewicht = np.array([6.46+6.46,9.6+4.52,48.91+6.46,50.84+6.46,120,150,245,535.9,551,603.16,604.36,486.23,367.77,247])
Arduino_waarde = np.array([63,102,272,150,510,650,675,875,823,855,856,867.6,814,740])

u_gewicht = 1
u_arduino = 20

def Functie(x,a,b):
    return a*np.log(x) + b

vals, cov = curve_fit(Functie,Gewicht,Arduino_waarde)

y_err = np.sqrt(cov[1,1])
x_err = np.sqrt(cov[0,0])

x_test= np.linspace(min(Gewicht),max(Gewicht),1000)
y_test= Functie(x_test, *vals)

plt.figure()
plt.plot(Gewicht,Arduino_waarde,"b.",label = "Meting")
plt.plot(x_test,y_test, label="Fitlijn")
plt.legend()
plt.xlabel("Gewicht($g$)")
plt.ylabel("Arduino_waarde")
plt.errorbar(Gewicht, Arduino_waarde,xerr=x_err,yerr=y_err,fmt='.')
plt.plot()

```
